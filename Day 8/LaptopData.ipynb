# Data Cleaning & Preprocessing

## Handling Missing Values

```python
# Drop completely empty trailing rows
df = df.dropna(how='all')

# Drop redundant index column if it exists
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])
```

```python
# Verify missing values count is zero
print("Missing values per column:")
print(df.isnull().sum()) 
```

## Removing Duplicate Records

```python
# Check and remove duplicate records 
initial_rows = len(df)
df = df.drop_duplicates()
print(f"Removed {initial_rows - len(df)} duplicate rows.")
print(f"Total rows remaining: {len(df)}")
```

## Summary of Data Cleaning Steps

- Every column now has 0 missing values. After handling missing values, the dataset is now complete and ready for analysis.
 
- Also removed 29 duplicate rows from the dataset. Now, the total rows remaining in the dataset are 1,244 which are unique &  high-quality records.

## Renaming Columns

```python
df = df.rename(columns={
    'Company': 'company',
    'TypeName': 'type_name',
    'Inches': 'inches',
    'ScreenResolution': 'screen_resolution',
    'Cpu': 'cpu',
    'Ram': 'ram',
    'Memory': 'memory',
    'Gpu': 'gpu',
    'OpSys': 'op_sys',
    'Weight': 'weight',
    'Price': 'price'
})

print("Updated column names:")
print(df.columns.tolist())
```

## Converting Incorrect Data Types

```python
# Convert columns to numeric
df['ram'] = df['ram'].astype(str).str.replace('GB', '', regex=False).str.strip()
df['ram'] = pd.to_numeric(df['ram'], errors='coerce')

df['weight'] = df['weight'].astype(str).str.replace('kg', '', regex=False).str.strip()
df['weight'] = pd.to_numeric(df['weight'], errors='coerce')

df['inches'] = pd.to_numeric(df['inches'], errors='coerce')


print(df.dtypes)
```

## Saving the Cleaned Dataset

```python
output_path = r"c:\Users\Aalhad Ramteke\Downloads\Cleaned_LaptopDataset.csv"
df.to_csv(output_path, index=False)

print(f"Cleaned dataset successfully saved to: {output_path}")
```

## Summary of Renaming Columns and Data Type Conversion

- All column headers were successfully standardized into lowercase (e.g., company, type_name, screen_resolution, op_sys) to ensure consistency for downstream analysis and machine learning pipelines.

- Several columns that represent quantitative metrics (ram, weight, and inches) were initially stored as string (str) object data types instead of numeric types.
