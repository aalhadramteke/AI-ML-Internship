## Data Filteration

```python
regions = df['Region'].unique() #displays the unique values in the 'Region' column
print(regions)
```

```python
# filters the dataset to only include rows where the 'Area' column is equal to 'Rural'
rural_data = df[df['Area'] == 'Rural'] # filters the dataset to only include rows where the 'Area' column is equal to 'Rural'
print(rural_data)
```

```python
# filters the dataset to only include rows where the 'Area' column is equal to 'Urban'
urban_data = df[df['Area'] == 'Urban'] 
print(urban_data)
```

```python
# Keep rows where the unemployment rate is greater than 50
df.columns = df.columns.str.strip()
df_high = df[df['Estimated Unemployment Rate (%)'] > 50] 
print(df_high)
print(df_high.shape) #displays the number of rows and columns in the filtered dataset
```

```python
# Keep rows where the unemployment rate is less than 30
df.columns = df.columns.str.strip()
df_high = df[df['Estimated Unemployment Rate (%)'] < 30] 
print(df_high)
print(df_high.shape) #displays the number of rows and columns in the filtered dataset
```
