# Importing Hugging Face Pipeline

```python
from transformers import pipeline
import warnings

warnings.filterwarnings("ignore", category=UserWarning)
print("Libraries imported successfully!")
```

# Defining Test Input Text

```python
sample_text = (
    "Artificial Intelligence and machine learning are transforming industries at an unprecedented pace. "
    "From healthcare diagnostics to automated financial auditing, companies that adopt AI see massive efficiency gains. "
    "However, this rapid growth brings critical challenges, including data privacy concerns, algorithmic bias, "
    "and the displacement of traditional jobs. Balancing innovation with ethical guardrails remains the defining challenge "
    "for developers and policymakers alike."
)
print("Input Text Loaded Successfully!")
```

# Test Model 1 – Text Generation
### Tests open-ended text completion using gpt2

```python
generator = pipeline("text-generation", model="gpt2")

gen_prompt = "The future of ethical AI development requires"
gen_output = generator(gen_prompt, max_length=500, num_return_sequences=1, truncation=True)

gen_output = generator(
    gen_prompt, 
    max_new_tokens=500,         # Control generation length safely
    num_return_sequences=1, 
    do_sample=True,             # Enable sampling for more creative text
    temperature=0.7,            # Controls randomness
    repetition_penalty=1.2      # Actively penalizes repeating phrases!
)
print("\n")
print("Text Generation Output:\n")
print(gen_output[0]['generated_text'])
```

# Test Model 2 – Sentiment Analysis
### Evaluates the emotional tone/sentiment of the input text using distilbert-base-uncased-finetuned-sst-2-english

```python
sentiment_classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

sentiment_output = sentiment_classifier(sample_text)

print("\n")
print("Sentiment Analysis Output")
print(f"Label: {sentiment_output[0]['label']}")
print(f"Confidence Score: {sentiment_output[0]['score']:.4f}")
```

# Test Model 3 – Summarization
### Generates a concise summary using sshleifer/distilbart-cnn-12-6.

```python
summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")

summary_output = summarizer(sample_text, max_length=60, min_length=25, do_sample=False)

print("Summarization Output")
print(summary_output[0]['summary_text'])
```

# Results Comparison & Performance Summary Table

```python
import pandas as pd

comparison_data = {
    "Task": ["Text Generation", "Sentiment Analysis", "Summarization"],
    "Model Name": [
        "gpt2",
        "distilbert-base-uncased-finetuned-sst-2-english",
        "sshleifer/distilbart-cnn-12-6"
    ],
    "Input Used": [gen_prompt, "Full sample_text", "Full sample_text"],
    "Output Summary": [
        gen_output[0]['generated_text'],
        f"{sentiment_output[0]['label']} ({sentiment_output[0]['score']*100:.1f}%)",
        summary_output[0]['summary_text']
    ],
    "Evaluation / Quality": [
        "Good continuation, requires length tuning",
        "Accurately detected overall positive/optimistic tone",
        "Concise, captured key risks and benefits accurately"
    ]
}

df_results = pd.DataFrame(comparison_data)
df_results
```

# Final Verdict

From my point of view, **`distilbert-base-uncased-finetuned-sst-2-english` (Sentiment Analysis)** performed best in terms of pure accuracy and output reliability.

* **High Confidence & Clarity:** It gave an exact, deterministic result with a **99.8% confidence score** (`POSITIVE`), leaving no ambiguity in its interpretation.
* **Zero Post-Processing Needed:** Unlike `gpt2` (which required tuning to prevent repetitive loops) or `distilbart` (which required specific length constraints), the sentiment classifier instantly delivered a precise, production-ready classification on the first try.
