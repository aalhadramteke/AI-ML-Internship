```python
import numpy as np
```

### Dataset Generation

```python
np.random.seed(10)

students = np.array([f"Student_{i:02d}" for i in range(1, 11)])

subjects = np.array(["Python","DSS","DBMS","ML","PDS"])

marks = np.random.randint(35, 100, size=(10, 5))

print("Students:", students.shape)
print("Subjects:", subjects)
print("Marks shape:", marks.shape)
print("\nFirst 5 students' marks:")
print(marks[:5])
```

### Basic EDA

```python
print("Array shape:", marks.shape)
print("Number of dimensions:", marks.ndim)
print("Data type:", marks.dtype)

print("\nTotal students:", marks.shape[0])
print("Total subjects:", marks.shape[1])
```

```python
print("First student's marks:")
print(marks[:1])

print("\nLast student's marks:")
print(marks[-1])
```

### Average Marks Calculation

```python
subject_averages = np.mean(marks, axis=0)

print("Subject-wise averages:")

for i in range(len(subjects)):
    print(f"{subjects[i]}: {subject_averages[i]:.2f}")
```

### Subject-wise Toppers

```python
print("SUBJECT-WISE TOPPERS")

for col in range(len(subjects)):
    topper_index = np.argmax(marks[:, col])
    topper_name = students[topper_index]
    topper_marks = marks[topper_index, col]

    print(
        f"{subjects[col]}: "
        f"{topper_name} scored {topper_marks}"
    )
```

### Students who failed

```python
failure_mask = np.any(marks < 40, axis=1)

failed_students = students[failure_mask]

print("Students with at least one subject below 40:")
print(failed_students)
print("Number of students who failed:", np.sum(failure_mask))
```

### Subject-wise passing report

```python
pass_mask = marks >= 40

passed_count = np.sum(pass_mask, axis=0)

pass_percentage = (passed_count / len(students)) * 100

print("SUBJECT-WISE PASSING REPORT")

for i in range(len(subjects)):
    print(
        f"{subjects[i]}: "
        f"{passed_count[i]}/10 passed, "
        f"Pass percentage = {pass_percentage[i]:.2f}%"
    )
```

## Student Marks Analysis

```python
print("STUDENT MARKS ANALYZER")

print(f"Total students: {len(students)}")
print(f"Total subjects: {len(subjects)}")
print(f"Total marks analyzed: {marks.size}")

print("\nSUBJECT-WISE REPORT")
for i, subject in enumerate(subjects):
    topper_index = np.argmax(marks[:, i])
    average = np.mean(marks[:, i])
    pass_rate = np.mean(marks[:, i] >= 40) * 100

    print(f"\nSubject: {subject}")
    print(f"Average: {average:.2f}")
    print(f"Topper: {students[topper_index]}")
    print(f"Top marks: {marks[topper_index, i]}")
    print(f"Pass percentage: {pass_rate:.2f}%")

print("\nSTUDENT PERFORMANCE")

print("Students with at least one failure:",
      np.sum(np.any(marks < 40, axis=1)))

print("Students scoring above 90 in at least one subject:",
      np.sum(np.any(marks > 90, axis=1)))
```
