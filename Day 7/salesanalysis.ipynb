## Changing Data Types

```python
# Changing the data type of discounted price and actual price

df['discounted_price'] = df['discounted_price'].str.replace("₹",'')
df['discounted_price'] = df['discounted_price'].str.replace(",",'')
df['discounted_price'] = df['discounted_price'].astype('float64')

df['actual_price'] = df['actual_price'].str.replace("₹",'')
df['actual_price'] = df['actual_price'].str.replace(",",'')
df['actual_price'] = df['actual_price'].astype('float64')
```

```python
# Changing Datatype and values in Discount Percentage

df['discount_percentage'] = df['discount_percentage'].str.replace('%','').astype('float64')

df['discount_percentage'] = df['discount_percentage'] / 100
```

```python
# Changing Rating Columns Data Type

df['rating'] = df['rating'].str.replace('|', '3.9').astype('float64')
```

```python
# Changing 'rating_count' Column Data Type

df['rating_count'] = df['rating_count'].str.replace(',', '').astype('float64')
```

## Data Visualization

### Chart 1: Distribution of Product Ratings

```python
plt.figure(figsize=(8, 4))
df['rating'].plot(kind='hist', bins=10, color='orange', edgecolor='black')
plt.title('Distribution of Product Ratings')
plt.xlabel('Rating Score')
plt.ylabel('Frequency')
plt.show()
```

### Chart 2: Top 5 Highest Priced Products (Discounted Price)

```python
plt.figure(figsize=(9, 4))
top_expensive = df.nlargest(5, 'discounted_price')
top_expensive.set_index(top_expensive['product_name'].str[:30])['discounted_price'].plot(kind='barh', color='teal')
plt.title('Top 5 Most Expensive Products')
plt.xlabel('Discounted Price')
plt.ylabel('Product Name')
plt.show()
```

### Chart 3: Top 5 Most Popular/Reviewed Products

```python
plt.figure(figsize=(9, 4))
top_reviewed = df.nlargest(5, 'rating_count')
top_reviewed.set_index(top_reviewed['product_name'].str[:25])['rating_count'].plot(kind='barh', color='mediumpurple')
plt.title('Top 5 Most Reviewed Products')
plt.xlabel('Total Rating Count (Number of Reviews)')
plt.ylabel('Product Name')
plt.show()
```

### Chart 4: Distribution of Discount Percentages (Histogram)

```python
plt.figure(figsize=(8, 4))
df['discount_percentage'].plot(kind='hist', bins=10, color='forestgreen', edgecolor='black')
plt.title('Distribution of Discount Percentages')
plt.xlabel('Discount Percentage (%)')
plt.ylabel('Number of Products')
plt.show()
```

### Chart 5: Average Actual Price vs. Discounted Price (Top 5 Categories)

```python
plt.figure(figsize=(9, 4))
df.groupby('category')[['actual_price', 'discounted_price']].mean().head(5).plot(kind='bar', figsize=(9, 4), color=['gray', 'dodgerblue'])
plt.title('Average Actual Price vs. Discounted Price by Category')
plt.xlabel('Category')
plt.ylabel('Price')
plt.xticks(rotation=45)
plt.legend(['Actual Price', 'Discounted Price'])
plt.show()
```
