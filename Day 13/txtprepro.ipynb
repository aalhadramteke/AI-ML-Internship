# Importing Libraries and NLTK Data

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

nltk.download('punkt') # Download the Punkt tokenizer models
nltk.download('stopwords') # Download the stopwords corpus

import warnings
warnings.filterwarnings("ignore")
print("Libraries imported successfully!")
```

# Defining a Sample Dataset

```python
corpus = [
    "I cannot stop smiling after hearing the incredible news about my promotion!",
"Sitting alone in the dark empty room made a deep wave of sorrow wash over her.",
"It is completely infuriating when people disregard your hard work and take all the credit.",
"My heart raced and cold sweat formed as the heavy footsteps echoed closer to the door.",
"I am utterly stunned that they planned such a massive surprise party without me suspecting anything.",
"Holding her newborn baby for the first time filled his soul with a profound and unconditional warmth.",
"The persistent failure of the system right before the deadline left everyone feeling completely helpless.",
"Walking through the misty morning forest paths always brings a sense of quiet calm to my mind.",
"How could you betray my trust like that after everything we have been through together?",
"The tightrope walker took a deep breath, steadying his balance as the crowd watched in absolute awe."
]

df = pd.DataFrame({'Original_Text': corpus})
display(df)
```

# Lowercasing the Data

```python
df['Lowercased'] = df['Original_Text'].str.lower()

display(df[['Original_Text', 'Lowercased']])
```

# Tokenization

```python
df['Tokens'] = df['Lowercased'].apply(word_tokenize)

display(df[['Lowercased', 'Tokens']])
```

# Stopword Removal

```python
# Removing stopwords and punctuation (keeping only alphanumeric words)

stop_words = set(stopwords.words('english'))

def remove_stopwords_and_punctuation(tokens):
    return [word for word in tokens if word.isalnum() and word not in stop_words]

df['Cleaned_Tokens'] = df['Tokens'].apply(remove_stopwords_and_punctuation)

# Rejoin tokens back into strings for the final text

df['Cleaned_Text'] = df['Cleaned_Tokens'].apply(lambda x: " ".join(x))

display(df[['Tokens', 'Cleaned_Text']])
```

# TF-IDF Vectorization

```python
# Applying TF-IDF Vectorization
vectorizer = TfidfVectorizer()

# Fit and transform the cleaned text corpus
tfidf_matrix = vectorizer.fit_transform(df['Cleaned_Text'])

# Get vocabulary (unique features/words)
feature_names = vectorizer.get_feature_names_out()

# Create a readable DataFrame of the TF-IDF matrix
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names)

print("Vocabulary Feature Names:")
print(feature_names)
print("\nTF-IDF Matrix DataFrame:")
display(tfidf_df)
```

###  **Final Summary**

1. **Dataset Creation:** We started with a small corpus of 10 raw sentences containing natural language text, punctuation, and capitalization.

2. **Lowercasing:** Standardized all text to lowercase so that words with different cases (e.g., *“Text”* vs. *“text”*) are treated identically by the computer.

3. **Tokenization:** Broke down each sentence into individual word units (tokens) to make the text analyzable at the granular word level.

4. **Stopword Removal & Cleaning:** Filtered out common, uninformative filler words (like *“is”*, *“an”*, *“in”*) and punctuation marks using NLTK's English stopword list, retaining only meaningful words.

5. **TF-IDF Vectorization:** Converted the cleaned text into a numerical matrix using Scikit-Learn. This computed **Term Frequency-Inverse Document Frequency (TF-IDF)** scores, transforming words into weighted numerical features that indicate how important a word is to a specific document relative to the entire dataset.
