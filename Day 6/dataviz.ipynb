## Data Visualization

```python
# Setting up a clean visual style
sns.set_theme(style="whitegrid")
```

```python
# Strip any accidental spaces from column names
df.columns = df.columns.str.strip()

# Verify the column exists
print(df.columns.tolist())
```

### Chart 1: Bar Chart (Average Balance by Geography)

```python
plt.figure(figsize=(7, 4))
df.groupby('Geography')['Balance'].mean().plot(kind='bar', color='skyblue')
plt.title('Average Balance by Geography')
plt.ylabel('Average Balance')
plt.show()
```

### Chart 2: Line Chart (Credit Score Trend)

```python
plt.figure(figsize=(10, 5))
plt.plot(df.index[:50], df['CreditScore'][:50], color='dodgerblue', linewidth=1.5)
plt.title("Credit Score Trend Across First 50 Customers", fontsize=14, fontweight='bold')
plt.xlabel("Customer Index", fontsize=12)
plt.ylabel("Credit Score", fontsize=12)
plt.show()
```

### Chart 3: Histogram (Age Distribution)

```python
plt.figure(figsize=(7, 4))
plt.hist(df['Age'], bins=20, color='orange', edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

### Chart 4: Pie Chart (Gender Proportion)

```python
plt.figure(figsize=(5, 5))
df['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Gender Proportion')
plt.ylabel('') 
plt.show()
```

## Key Insights

### Geographic Balances: The bar chart shows that customers in Germany have a much higher average bank balance compared to France and Spain.

### Credit Scores: The line chart tracking the first 50 customers shows typical fluctuations in credit scores ranging mostly between 500 and 850.

### Age Spread: The histogram highlights that customer ages are concentrated heavily between 30 and 45 years old.

### Gender Split: The pie chart reveals a fairly balanced split between male and female customers in the bank's database, with a slightly higher percentage of male customers.
